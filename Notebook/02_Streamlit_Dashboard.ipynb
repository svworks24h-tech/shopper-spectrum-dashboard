# Executive Dashboard

```python
sales_df.columns.tolist()
```

```python
sales_df["TotalPrice"] = sales_df["Quantity"] * sales_df["UnitPrice"]
```

```python
# ============================================================
# Executive Dashboard KPIs
# ============================================================

total_revenue = sales_df["TotalPrice"].sum()

total_customers = sales_df["CustomerID"].nunique()

total_orders = sales_df["InvoiceNo"].nunique()

total_products = sales_df["Description"].nunique()

print("Revenue :", total_revenue)
print("Customers :", total_customers)
print("Orders :", total_orders)
print("Products :", total_products)
```

# Prepare Dashboard Data

```python
# Monthly Sales

sales_df["InvoiceDate"] = pd.to_datetime(sales_df["InvoiceDate"])

sales_df["Month"] = sales_df["InvoiceDate"].dt.to_period("M").astype(str)

monthly_sales = sales_df.groupby("Month")["TotalPrice"].sum().reset_index()

top_products = (
    sales_df.groupby("Description")["Quantity"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

country_sales = (
    sales_df.groupby("Country")["TotalPrice"]
    .sum()
    .sort_values(ascending=False)
)
```
